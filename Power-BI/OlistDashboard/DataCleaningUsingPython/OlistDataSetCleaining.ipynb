```python
import re
import html

def clean_city(text):
    if isinstance(text, str):
        # Decode HTML entities
        text = html.unescape(text)

        # Remove unwanted symbols
        text = re.sub(r'[^A-Za-zÀ-ÿ0-9\s\(\)-]', '', text)

        # Clean extra spaces
        text = re.sub(r'\s+', ' ', text).strip()

        return text
    
    return text


geo['geolocation_city'] = geo['geolocation_city'].apply(clean_city)
```

```python
import re

def find_cities_with_numbers(df, column='customer_city'):
    """
    Returns rows where the specified column contains one or more digits.
    """
    return df[df[column].astype(str).str.contains(r'\d', regex=True, na=False)]

cities_with_numbers = find_cities_with_numbers(customers)

print(cities_with_numbers)
```

```python
import re

def find_cities_with_special_symbols(df, column='customer_city'):
    """
    Returns rows where the city name contains special symbols.
    Keeps letters (including accented letters), numbers, and spaces.
    """
    pattern = r'[^A-Za-zÀ-ÿ0-9\s]'

    return df[df[column].astype(str).str.contains(pattern, regex=True, na=False)]


special_cities = find_cities_with_special_symbols(customers)
print(special_cities['customer_city'])
```

```python
import re

def clean_city_special_symbols(df, column='customer_city'):
    """
    Removes unwanted special symbols while keeping:
    - letters
    - accented letters
    - numbers
    - spaces
    - apostrophes (')
    - hyphens (-)
    """

    df[column] = (
        df[column]
        .astype(str)
        .str.replace(r"[^A-Za-zÀ-ÿ0-9\s'\-]", "", regex=True)
        .str.strip()
    )

    return df
```

```python
customers = clean_city_special_symbols(customers)
print(customers['customer_city'].head())
```
